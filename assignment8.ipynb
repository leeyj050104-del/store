## 1. 해싱 충돌 해결 방법 비교

해싱에서 충돌(Collision)이 발생했을 때 해결하는 대표적인 방법은 크게 **개방 주소법(Open Addressing)** 과 **체이닝(Chaining)** 이다.

| 구분 | 개방 주소법(Open Addressing) | 체이닝(Chaining) |
|---|---|---|
| 방법 | 충돌 발생 시 해시 테이블 안에서 다른 빈 공간을 찾아 저장 | 같은 해시 주소에 연결 리스트로 여러 키를 저장 |
| 장점 | 추가 메모리가 거의 필요 없음 | 충돌이 많이 발생해도 저장 가능 |
| 장점 | 캐시 효율이 좋아 탐색 속도가 빠를 수 있음 | 테이블이 꽉 차도 연결 리스트로 계속 저장 가능 |
| 단점 | 충돌이 많아지면 탐색 시간이 급격히 증가 | 연결 리스트를 위한 추가 메모리가 필요 |
| 단점 | 테이블이 거의 차면 성능 저하가 심함 | 연결 리스트 탐색 때문에 캐시 효율이 낮음 |
| 특징 | 클러스터링 문제가 발생할 수 있음 | 삽입과 삭제가 비교적 쉬움 |

## 01. 해시 테이블 크기와 해시 함수의 문제점

해시 테이블 크기는 1,204인데 해시 함수로 다음을 사용하였다.

```text
h(x) = x % 1024
```

문제점은 해시 함수의 결과값이 **0부터 1023까지만** 나온다는 것이다.
즉, 해시 테이블의 인덱스 **1024부터 1203까지는 절대 사용되지 않는다.**

따라서 다음과 같은 결점이 있다.

- 테이블 공간이 낭비된다.
- 데이터가 일부 구간에만 집중된다.
- 충돌 가능성이 증가한다.
- 해시 테이블 전체에 균등하게 분포되지 않는다.

즉, 해시 함수에서 나누는 값과 해시 테이블의 크기가 일치하지 않는 것이 문제이다.

## 02. random을 사용한 해시 함수가 될 수 없는 이유

다음 함수는 해시 함수가 될 수 없다.

```text
h(x) = x * random(0,10) % m
```

그 이유는 같은 키 `x`가 입력되어도 `random(0,10)`의 값이 매번 달라지기 때문이다.

예를 들어 같은 키 `10`을 넣어도 어떤 순간에는 `h(10)=3`, 다른 순간에는 `h(10)=7`처럼 결과가 바뀔 수 있다.

해시 함수는 반드시 같은 입력에 대해 항상 같은 결과를 반환해야 한다. 그래야 데이터를 저장한 위치를 다시 찾아갈 수 있다.

하지만 위 함수는 결과가 매번 달라질 수 있으므로 저장된 데이터를 검색할 수 없다. 따라서 올바른 해시 함수가 될 수 없다.

## 07. 체이닝에서 맨 뒤에 삽입할 경우 효율성 변화

체이닝을 사용하는 충돌 해결 방법에서 새로운 키를 연결 리스트의 맨 앞이 아니라 맨 뒤에 삽입하면 다음과 같은 변화가 생긴다.

장점

- 데이터가 삽입된 순서를 유지할 수 있다.

단점

- 맨 앞 삽입은 바로 넣으면 되므로 O(1)이다.
- 맨 뒤 삽입은 마지막 노드까지 이동해야 하므로 O(n)이 될 수 있다.

따라서 검색 성능은 큰 차이가 없지만, 삽입 효율은 감소한다.

| 연산 | 효율 변화 | 이유 |
|---|---|---|
| 삽입 | 느려짐 | 정렬된 위치를 찾아 삽입해야 하기 때문 |
| 삭제 | 경우에 따라 좋아질 수 있음 | 정렬되어 있으면 찾는 과정에서 중간 종료가 가능하기 때문 |
| 성공적인 검색 | 약간 좋아질 수 있음 | 정렬된 순서대로 비교하므로 불필요한 탐색이 줄어들 수 있음 |
| 실패하는 검색 | 좋아짐 | 찾는 값보다 큰 값이 나오면 더 이상 탐색하지 않아도 되기 때문 |
